```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder

from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LogisticRegression

from sklearn.metrics import accuracy_score, classification_report


# Load dataset
data = pd.read_csv("Z:\python\weather_dataset.csv")


# Separate input and output
x = data.drop("Weather_Condition", axis=1)
y = data["Weather_Condition"]


# Encode weather conditions
label_encoder = LabelEncoder()
y = label_encoder.fit_transform(y)


# Split dataset
x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)


# ==================================================
# DECISION TREE
# ==================================================

dt = DecisionTreeClassifier(random_state=42)

dt.fit(x_train, y_train)

dt_pred = dt.predict(x_test)

dt_accuracy = accuracy_score(y_test, dt_pred)


# ==================================================
# RANDOM FOREST
# ==================================================

rf = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

rf.fit(x_train, y_train)

rf_pred = rf.predict(x_test)

rf_accuracy = accuracy_score(y_test, rf_pred)


# ==================================================
# KNN
# ==================================================

knn = KNeighborsClassifier(n_neighbors=5)

knn.fit(x_train, y_train)

knn_pred = knn.predict(x_test)

knn_accuracy = accuracy_score(y_test, knn_pred)


# ==================================================
# LOGISTIC REGRESSION
# ==================================================

lr = LogisticRegression(max_iter=1000)

lr.fit(x_train, y_train)

lr_pred = lr.predict(x_test)

lr_accuracy = accuracy_score(y_test, lr_pred)


# ==================================================
# ACCURACY
# ==================================================

print("\nMODEL ACCURACY")

print("Decision Tree:", dt_accuracy)

print("Random Forest:", rf_accuracy)

print("KNN:", knn_accuracy)

print("Logistic Regression:", lr_accuracy)


# ==================================================
# CLASSIFICATION REPORT - DECISION TREE
# ==================================================

print("\n==========================================")
print("DECISION TREE CLASSIFICATION REPORT")
print("==========================================")

print(classification_report(y_test, dt_pred))


# ==================================================
# CLASSIFICATION REPORT - RANDOM FOREST
# ==================================================

print("\n==========================================")
print("RANDOM FOREST CLASSIFICATION REPORT")
print("==========================================")

print(classification_report(y_test, rf_pred))


# ==================================================
# CLASSIFICATION REPORT - KNN
# ==================================================

print("\n==========================================")
print("KNN CLASSIFICATION REPORT")
print("==========================================")

print(classification_report(y_test, knn_pred))


# ==================================================
# CLASSIFICATION REPORT - LOGISTIC REGRESSION
# ==================================================

print("\n==========================================")
print("LOGISTIC REGRESSION CLASSIFICATION REPORT")
print("==========================================")

print(classification_report(y_test, lr_pred))


# ==================================================
# FOUR MODELS COMPARISON
# ==================================================

results = pd.DataFrame({
    "Model": [
        "Decision Tree",
        "Random Forest",
        "KNN",
        "Logistic Regression"
    ],
    "Accuracy": [
        dt_accuracy,
        rf_accuracy,
        knn_accuracy,
        lr_accuracy
    ]
})

print("\n==========================================")
print("FOUR MODELS COMPARISON")
print("==========================================")

print(results.to_string(index=False))


# ==================================================
# BAR CHART
# ==================================================

models = [
    "Decision Tree",
    "Random Forest",
    "KNN",
    "Logistic Regression"
]

accuracies = [
    dt_accuracy,
    rf_accuracy,
    knn_accuracy,
    lr_accuracy
]

plt.figure(figsize=(8, 5))

plt.bar(models, accuracies)

plt.xlabel("Models")
plt.ylabel("Accuracy")

plt.title("Weather Prediction Model Comparison")

plt.xticks(rotation=15)

plt.show()




# User Input Prediction

temperature = float(input("Temperature: "))
humidity = float(input("Humidity: "))
pressure = float(input("Pressure: "))
wind_speed = float(input("Wind Speed: "))
rainfall = float(input("Rainfall: "))

input_data = [[temperature, humidity, pressure, wind_speed, rainfall]]

prediction = lr.predict(input_data)

print("Predicted Weather:", label_encoder.inverse_transform(prediction)[0])
```
